**Objetivos:**

- Demostrar el principio de desacoplamiento y polimorfismo mediante una función cliente genérica.


### 1. Definición del ADT de Pila (`ADTPila`)
Definimos el contrato formal usando clases abstractas (`ABC`) y `@abstractmethod`.


```python
from abc import ABC, abstractmethod

class ADTPila(ABC):
    """Interfaz abstracta de una pila (LIFO)."""

    @abstractmethod
    def push(self, item):
        """Inserta un elemento en la cima."""
        pass

    @abstractmethod
    def pop(self):
        """Elimina y devuelve el elemento en la cima."""
        pass

    @abstractmethod
    def top(self):
        """Devuelve el elemento en la cima sin eliminarlo."""
        pass

    @abstractmethod
    def is_empty(self) -> bool:
        """Indica si la pila está vacía."""
        pass

    @abstractmethod
    def __len__(self) -> int:
        """Número de elementos en la pila."""
        pass
```

### 2. Implementación 1: Pila con Arreglo Dinámico (`PilaArray`)
Utiliza la estructura interna de arreglo redimensionable de Python (`list`).


```python
class PilaArray(ADTPila):
    """Implementación de ADTPila basada en un arreglo dinámico (list)."""

    def __init__(self):
        self._datos = []

    def push(self, item):
        self._datos.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("Pila vacía")
        return self._datos.pop()
    def top(self):
        if self.is_empty():
            raise IndexError("Pila vacía")
        return self._datos[-1]

    def is_empty(self) -> bool:
        return len(self._datos) == 0

    def __len__(self) -> int:
        return len(self._datos)
```

### 3. Implementación 2: Pila con Lista Enlazada Simple (`PilaLista`)
Utiliza una cadena dinámica de nodos (`_NodoPila`) conectados mediante referencias (`siguiente`).


```python
class _NodoPila:
    __slots__ = ("dato", "siguiente")

    def __init__(self, dato, siguiente=None):
        self.dato = dato
        self.siguiente = siguiente


class PilaLista(ADTPila):
    """Implementación de ADTPila basada en lista enlazada simple."""

    def __init__(self):
        self._tope = None
        self._n = 0

    def push(self, item):
        self._tope = _NodoPila(item, self._tope)
        self._n += 1

    def pop(self):
        if self.is_empty():
            raise IndexError("Pila vacía")
        dato = self._tope.dato
        self._tope = self._tope.siguiente
        self._n -= 1
        return dato

    def top(self):
        if self.is_empty():
            raise IndexError("Pila vacía")
        return self._tope.dato

    def is_empty(self) -> bool:
        return self._tope is None

    def __len__(self) -> int:
        return self._n
```
