### Ejercicio 4: Modelos de Persistencia (JSON y Pickle)

**Objetivos:**
- Implementar mecanismos de persistencia en almacenamiento secundario para convertir estructuras de datos en memoria en representaciones durables.
- Comparar dos formatos fundamentales en Python:
  1. **JSON (JavaScript Object Notation):** Formato de texto estructurado, interoperable y legible por humanos.
  2. **Pickle:** Serialización binaria nativa de Python para objetos complejos.
- Validar la integridad de los datos guardados y recuperados mediante aserciones (`assert`).


### 1. Definición de Entidad y Generación de Datos


```python
from dataclasses import dataclass, asdict
import json
import os
import pickle
import random

@dataclass
class Estudiante:
    nombre: str
    nota: float
    grupo: str

def generar_estudiantes(n=10, semilla=42):
    random.seed(semilla)
    nombres = ["Ana", "Luis", "María", "Carlos", "Elena", "Pedro", "Sofía", "Diego"]
    return [
        Estudiante(
            nombre=random.choice(nombres),
            nota=round(random.uniform(1.0, 7.0), 1),
            grupo=random.choice(["A", "B", "C"]),
        )
        for _ in range(n)
    ]

estudiantes_muestra = generar_estudiantes(5)
for e in estudiantes_muestra:
    print(e)
```

### 2. Funciones de Persistencia JSON y Pickle


```python
# --- Persistencia JSON ---
def guardar_json(estudiantes, archivo):
    """Serializa la lista de dataclasses a formato JSON de texto plano."""
    with open(archivo, "w", encoding="utf-8") as f:
        json.dump([asdict(e) for e in estudiantes], f, ensure_ascii=False, indent=2)

def cargar_json(archivo):
    """Deserializa un archivo JSON y reconstruye las instancias de Estudiante."""
    with open(archivo, "r", encoding="utf-8") as f:
        datos = json.load(f)
    return [Estudiante(**d) for d in datos]

# --- Persistencia Pickle ---
def guardar_pickle(estudiantes, archivo):
    """Serializa la lista de objetos en un flujo de bytes binarios."""
    with open(archivo, "wb") as f:
        pickle.dump(estudiantes, f)

def cargar_pickle(archivo):
    """Deserializa el flujo binario reconstruyendo los objetos en memoria."""
    with open(archivo, "rb") as f:
        return pickle.load(f)
```

### 3. Verificación de Integridad y Lectura de Archivos


```python
tmp_json = "estudiantes_test.json"
tmp_pkl = "estudiantes_test.pkl"

# Guardar en ambos formatos
guardar_json(estudiantes_muestra, tmp_json)
guardar_pickle(estudiantes_muestra, tmp_pkl)

# Cargar y verificar integridad
recuperados_json = cargar_json(tmp_json)
recuperados_pkl = cargar_pickle(tmp_pkl)

assert recuperados_json == estudiantes_muestra, "JSON: los datos recuperados no coinciden"
assert recuperados_pkl == estudiantes_muestra, "Pickle: los datos recuperados no coinciden"

print("✓ JSON guardado y recuperado correctamente:")
with open(tmp_json, "r", encoding="utf-8") as f:
    print(f.read())

print("✓ Pickle guardado y recuperado correctamente:")
print(f"Tamaño binario en disco: {os.path.getsize(tmp_pkl)} bytes")

# Limpieza
if os.path.exists(tmp_json): os.remove(tmp_json)
if os.path.exists(tmp_pkl): os.remove(tmp_pkl)
```

### 4. Cuadro Comparativo: JSON vs Pickle

| Criterio | JSON (`.json`) | Pickle (`.pkl`) |
| :--- | :--- | :--- |
| **Naturaleza** | Texto plano estructurado (UTF-8) | Flujo binario de bytes |
| **Legibilidad** | Alta (legible y editable directamente por humanos) | Nula (código binario opaco) |
| **Interoperabilidad** | Universal (soportado por cualquier lenguaje o API) | Exclusivo del ecosistema de Python |
| **Soporte de tipos** | Tipos primitivos (números, cadenas, listas, diccionarios) | Objetos Python complejos, funciones, clases |
| **Seguridad** | Seguro (solo datos) | Inseguro si se cargan fuentes no confiables (ejecución de código arbitrario) |
