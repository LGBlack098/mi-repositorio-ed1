## Unidad I: Modelos de Representación de Datos
### Ejercicio 2: Modelo Estático vs. Modelo Dinámico

**Objetivos:**
- Implementar una estructura de representación estática (`ArrayEstatico`) con capacidad fija y control explícito de desbordamiento (*overflow*).
- Implementar una estructura de representación dinámica (`ListaDinamica`) basada en nodos enlazados con punteros a cabeza y cola.
- Analizar y comparar la complejidad algorítmica ($O$) de las operaciones fundamentales en ambos modelos.


### 1. Estructura Estática: `ArrayEstatico`
Capacidad fija reservada en memoria desde la inicialización.


```python
class ArrayEstatico:
    """Arreglo de capacidad fija con control de desbordamiento."""

    def __init__(self, capacidad):
        if capacidad <= 0:
            raise ValueError("La capacidad debe ser positiva")
        self._capacidad = capacidad
        self._datos = [None] * capacidad
        self._n = 0

    def insertar_final(self, item):
        if self._n >= self._capacidad:
            raise OverflowError("Desbordamiento: capacidad máxima alcanzada")
        self._datos[self._n] = item
        self._n += 1

    def obtener(self, indice):
        if indice < 0 or indice >= self._n:
            raise IndexError("Índice fuera de rango")
        return self._datos[indice]

    def eliminar(self, indice):
        if indice < 0 or indice >= self._n:
            raise IndexError("Índice fuera de rango")
        for i in range(indice, self._n - 1):
            self._datos[i] = self._datos[i + 1]
        self._datos[self._n - 1] = None
        self._n -= 1

    def __len__(self):
        return self._n

    def __iter__(self):
        for i in range(self._n):
            yield self._datos[i]
```

### 2. Estructura Dinámica: `ListaDinamica`
Crecimiento bajo demanda nodo a nodo sin límite fijo preasignado.


```python
class _Nodo:
    __slots__ = ("dato", "siguiente")

    def __init__(self, dato, siguiente=None):
        self.dato = dato
        self.siguiente = siguiente


class ListaDinamica:
    """Lista enlazada simple con inserción O(1) al inicio y al final."""

    def __init__(self):
        self._cabeza = None
        self._cola = None
        self._n = 0

    def insertar_inicio(self, item):
        nuevo = _Nodo(item, self._cabeza)
        self._cabeza = nuevo
        if self._cola is None:
            self._cola = nuevo
        self._n += 1
    def insertar_final(self, item):
        nuevo = _Nodo(item)
        if self._cola is None:
            self._cabeza = self._cola = nuevo
        else:
            self._cola.siguiente = nuevo
            self._cola = nuevo
        self._n += 1

    def obtener(self, indice):
        if indice < 0 or indice >= self._n:
            raise IndexError("Índice fuera de rango")
        actual = self._cabeza
        for _ in range(indice):
            actual = actual.siguiente
        return actual.dato

    def eliminar(self, indice):
        if indice < 0 or indice >= self._n:
            raise IndexError("Índice fuera de rango")
        if indice == 0:
            self._cabeza = self._cabeza.siguiente
            if self._cabeza is None:
                self._cola = None
        else:
            anterior = self._cabeza
            for _ in range(indice - 1):
                anterior = anterior.siguiente
            eliminado = anterior.siguiente
            anterior.siguiente = eliminado.siguiente
            if eliminado is self._cola:
                self._cola = anterior
        self._n -= 1

    def eliminar_inicio(self):
        self.eliminar(0)

    def __len__(self):
        return self._n

    def __iter__(self):
        actual = self._cabeza
        while actual is not None:
            yield actual.dato
            actual = actual.siguiente
```

### 3. Pruebas y Demostración de Operaciones


```python
def probar_estatico_dinamico():
    print("--- ArrayEstatico ---")
    arr = ArrayEstatico(3)
    arr.insertar_final("a")
    arr.insertar_final("b")
    arr.insertar_final("c")
    print("Contenido inicial:", list(arr))
    try:
        arr.insertar_final("d")
    except OverflowError as e:
        print("Desbordamiento controlado ->", e)
    arr.eliminar(1)
    print("Tras eliminar índice 1:", list(arr))

    print("\n--- ListaDinamica ---")
    lst = ListaDinamica()
    lst.insertar_final(1)
    lst.insertar_final(2)
    lst.insertar_inicio(0)
    print("Contenido inicial:", list(lst))
    lst.eliminar_inicio()
    print("Tras eliminar_inicio:", list(lst))

probar_estatico_dinamico()
```

### 4. Tabla Comparativa de Complejidad Asintótica

| Operación | `ArrayEstatico` | `ListaDinamica` (enlazada) |
| :--- | :--- | :--- |
| **Inserción al final** | $O(1)$ (si hay espacio) | $O(1)$ (con referencia a cola) |
| **Inserción al inicio** | $O(n)$ (requiere desplazar elementos) | $O(1)$ (cambio de puntero cabeza) |
| **Acceso por índice** | $O(1)$ (acceso directo por cálculo de offset) | $O(n)$ (recorrido secuencial de nodos) |
| **Eliminación al inicio** | $O(n)$ (desplazamiento a la izquierda) | $O(1)$ (avance de puntero cabeza) |
| **Eliminación por índice** | $O(n)$ (desplazamiento) | $O(n)$ (búsqueda secuencial del nodo) |
| **Búsqueda por valor** | $O(n)$ | $O(n)$ |
